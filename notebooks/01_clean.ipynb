# Student Engagement and Performance: Data Cleaning
Goal: clean the xAPI-Edu-Data dataset for a Power BI dashboard and an early-warning model.
Data: public dataset from Kaggle (480 records, 17 columns).

```python
# Cell 1: Load libraries and read the raw data
import pandas as pd

RAW_PATH = "../data/raw/xAPI-Edu-Data.csv"

df_raw = pd.read_csv(RAW_PATH)

print("Rows, columns:", df_raw.shape)
df_raw.head()
```

# Check data quality

```python
# Cell 2: Check data types, missing values and duplicates
df_raw.info()

print("\nMissing values in total:", df_raw.isna().sum().sum())
print("Exact duplicate rows:", df_raw.duplicated().sum())

# Show every row that has an exact copy
df_raw[df_raw.duplicated(keep=False)]
```

```python
# Cell 3: Look at every value in the text columns to find typos
text_cols = df_raw.select_dtypes(include="object").columns

for col in text_cols:
    print(f"--- {col} ({df_raw[col].nunique()} values)")
    print(df_raw[col].value_counts().to_dict())
```

## 3. Clean the data
Removed 2 exact duplicate rows (rows 326 and 327 copy rows 322 and 323), then fixed column name typos.

```python
# Cell 4: Work on a copy, remove exact duplicates and fix column names
df = df_raw.drop_duplicates().copy()

df = df.rename(columns={
    "gender": "Gender",
    "NationalITy": "Nationality",
    "PlaceofBirth": "Place_of_Birth",
    "StageID": "Stage",
    "GradeID": "Grade",
    "SectionID": "Section",
    "raisedhands": "Raised_Hands",
    "VisITedResources": "Visited_Resources",
    "AnnouncementsView": "Announcements_Viewed",
    "Discussion": "Discussion_Posts",
    "ParentAnsweringSurvey": "Parent_Answered_Survey",
    "ParentschoolSatisfaction": "Parent_Satisfaction",
    "StudentAbsenceDays": "Absence_Days",
    "Class": "Class_Code",
})

print("Rows before:", len(df_raw), "| Rows after:", len(df))
print(list(df.columns))
```

```python
# Cell 5: Standardize country names and make coded values easy to read
country_fix = {
    "KW": "Kuwait",
    "KuwaIT": "Kuwait",
    "lebanon": "Lebanon",
    "venzuela": "Venezuela",
    "Lybia": "Libya",
    "Tunis": "Tunisia",
    "SaudiArabia": "Saudi Arabia",
}
df["Nationality"] = df["Nationality"].replace(country_fix)
df["Place_of_Birth"] = df["Place_of_Birth"].replace(country_fix)

df["Stage"] = df["Stage"].replace({
    "lowerlevel": "Elementary",
    "MiddleSchool": "Middle School",
    "HighSchool": "High School",
})

df["Semester"] = df["Semester"].replace({"F": "First", "S": "Second"})

for col in ["Nationality", "Place_of_Birth", "Stage", "Semester"]:
    print(f"--- {col}")
    print(df[col].value_counts().to_dict())
```

## 4. Add helper columns
Engagement_Score is the simple average of the four engagement measures (0–100). It is not an original field.

```python
# Cell 6: Add helper columns for the dashboard and the model
df = df.reset_index(drop=True)
df.insert(0, "Record_ID", df.index + 1)

df["Grade_Number"] = df["Grade"].str.replace("G-", "").astype(int)

df["Performance"] = df["Class_Code"].map({"L": "Low", "M": "Medium", "H": "High"})
df["Performance_Order"] = df["Class_Code"].map({"L": 1, "M": 2, "H": 3})
df["Is_Low_Performer"] = df["Class_Code"] == "L"

df["Absent_7_Plus"] = df["Absence_Days"] == "Above-7"

engagement_cols = ["Raised_Hands", "Visited_Resources", "Announcements_Viewed", "Discussion_Posts"]
df["Engagement_Score"] = df[engagement_cols].mean(axis=1).round(1)

df[["Record_ID", "Grade", "Grade_Number", "Class_Code", "Performance",
    "Is_Low_Performer", "Absence_Days", "Absent_7_Plus", "Engagement_Score"]].head()
```

## 5. Validate and save
Automatic checks run before saving, so the clean file always matches the rules above.

```python
# Cell 7: Run final checks, then save the clean data
CLEAN_PATH = "../data/clean/student_engagement_clean.csv"

# 1. Row and column counts
assert len(df) == 478, "Row count should be 478 after removing duplicates"
assert df["Record_ID"].is_unique, "Record_ID must be unique"

# 2. No missing values and no duplicates left
assert df.isna().sum().sum() == 0, "There should be no missing values"
assert not df.drop(columns="Record_ID").duplicated().any(), "Duplicates remain"

# 3. Engagement values stay between 0 and 100
for col in engagement_cols + ["Engagement_Score"]:
    assert df[col].between(0, 100).all(), f"{col} has values outside 0-100"

# 4. Old typos are gone
old_values = ["KW", "KuwaIT", "lebanon", "venzuela", "Lybia", "Tunis", "SaudiArabia"]
assert not df["Nationality"].isin(old_values).any(), "Nationality typos remain"
assert not df["Place_of_Birth"].isin(old_values).any(), "Place_of_Birth typos remain"

print("All checks passed.")
print("Performance counts:", df["Performance"].value_counts().to_dict())
print("Stage vs grade mismatch rows:",
      len(df[(df["Stage"] == "Elementary") & (df["Grade_Number"] > 5)]))

df.to_csv(CLEAN_PATH, index=False)
print("Saved:", CLEAN_PATH, df.shape)
```
